# WF-001 Stage-1 Walkthrough (Notebook)

This notebook is a **step-by-step demonstration** of the WF-001 stage-1 pipeline.

It runs:

- the config-driven demo benchmark (`scripts/run_demo.py`)
- the report builder (`scripts/build_report.py`)
- and then previews the aggregate table and summary.

> Note: The primary (canonical) path is still the CLI. This notebook is just a guided demo.


```python
from pathlib import Path

REPO_ROOT = Path('..').resolve()
RUNS_DIR = REPO_ROOT / 'data' / 'outputs' / 'runs'
REPORT_DIR = REPO_ROOT / 'data' / 'outputs' / 'reports'

print('Repo root:', REPO_ROOT)
print('Runs dir:', RUNS_DIR)
print('Report dir:', REPORT_DIR)
```

```python
import subprocess

# Run the demo (config-driven)
subprocess.run(
    [
        'python', str(REPO_ROOT / 'scripts' / 'run_demo.py'),
        '--config', str(REPO_ROOT / 'configs' / 'demo' / 'wf001_demo_autompg.yaml'),
    ],
    check=True,
)
```

```python
# Build the report
subprocess.run(
    [
        'python', str(REPO_ROOT / 'scripts' / 'build_report.py'),
        '--input-root', str(RUNS_DIR),
        '--output-root', str(REPORT_DIR),
    ],
    check=True,
)
```

```python
import pandas as pd

summary_csv = REPORT_DIR / 'summary_table.csv'
summary_md = REPORT_DIR / 'demo_summary.md'
fig_png = REPORT_DIR / 'eps_vs_test_mse.png'

display(pd.read_csv(summary_csv))
print('\n--- demo_summary.md ---\n')
print(summary_md.read_text())
print('\nFigure path:', fig_png)
```
